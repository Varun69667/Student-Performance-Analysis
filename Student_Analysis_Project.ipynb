```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

print("Student performance Analysis")
print("Libraries imported successfully")
```

```python
data = {
    "Name" : ["Rahul","Varun","Ravi"],
    "Maths" : [70,80,90],
    "Science" : [67,69,80],
    "English" : [78,90,91],
    "Attendance" : [92,89,65]
}

df = pd.DataFrame(data)
print(df)
```

```python
df["Average"] = (
    df["Maths"] + df["Science"] + df["English"]
) / 3

print(df[["Name","Average"]])
```

```python
plt.bar(df["Name"],df["Average"])
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.title("Student Performance")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Attendance"],df["Average"])
plt.title("Attendance vs Student Performance")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")
plt.grid()
plt.show()
```

```python
correlation = df["Attendance"].corr(df["Average"])
print("Attendance and Marks Correlation:",correlation)

if correlation > 0:
  print("Higher attendance is associated with higher marks")

elif correlation < 0:
  print("Higher attendance is associated with lower marks")

else:
  print("No Linear correlation found")
```

```python
def get_grade(marks):
  if marks >= 90:
    return "A"

  elif marks >= 80:
    return "B"

  elif marks >= 70:
    return "C"

  else:
    return "D"

df["Grade"] = df["Average"].apply(get_grade)
print("Student Grades:")
print(df[["Name","Average","Grade"]])
```

```python
print("STUDENT PERFORMANCE REPORT")
print("--------------------------")

print("Total students:",len(df))

print("Class Average:",round(df["Average"].mean(),2))

print("Average Attendance:",round(df["Attendance"].mean(),2),"%")

print("Top Performer:",
      df.loc[df["Average"].idxmax(),"Name"]
)

print("Bottom Performer:",
      df.loc[df["Average"].idxmin(),"Name"]
)
```

```python

new_students = pd.DataFrame({
    "Name": ["Aarav", "Priya", "Sneha", "Kiran",
             "Ananya", "Arjun", "Meera"],
    "Maths": [85, 95, 88, 76, 92, 90, 81],
    "Science": [90, 93, 85, 80, 96, 89, 84],
    "English": [78, 91, 90, 73, 88, 92, 85],
    "Attendance": [92, 98, 95, 78, 96, 94, 88]
})

df = pd.concat([df, new_students], ignore_index=True)

df["Total"] = df["Maths"] + df["Science"] + df["English"]
df["Average"] = df["Total"] / 3

df["Grade"] = df["Average"].apply(get_grade)

print(df)
print("\nTotal students:", len(df))
```

```python

plt.figure(figsize=(12, 5))

plt.bar(df["Name"], df["Average"])

plt.title("Student Performance Analysis")
plt.xlabel("Student Names")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

grade_count = df["Grade"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(grade_count.index, grade_count.values)

plt.title("Grade Distribution")
plt.xlabel("Grades")
plt.ylabel("Number of Students")

plt.show()
```

```python

print("FINAL PROJECT INSIGHTS")
print("----------------------")

print("Total Students:", len(df))

print("Highest Average:",
      round(df["Average"].max(), 2))

print("Lowest Average:",
      round(df["Average"].min(), 2))

print("Class Average:",
      round(df["Average"].mean(), 2))

print("\nGrade Distribution:")
print(df["Grade"].value_counts())

print("\nAttendance Correlation:",
      round(df["Attendance"].corr(df["Average"]), 2))
```

```python

df.to_csv("student_performance.csv", index=False)

print("Dataset saved successfully!")
```
